```python
def rmsd_ref(path, prot1, prot2, file, mutation=False):
    """
    Parameters:
        path (str):
        prot1 (str):
        prot2 (str):
        file (str):
        mutation (bool):
    """
    x, y = [], []
    
    # Open the GROMAC file
    with open(path+file) as f:
        for line in f:
            if line.startswith(("@", "#")):
                continue
            cols = line.strip().split()
            x.append(float(cols[0])/1000) # time in ns
            y.append(float(cols[1])*10) # RMSD in Å
    
    mean_y = np.mean(y)
    std_y = np.std(y)
    
    # Plot
    plt.figure(figsize=(12, 6))
    plt.plot(x, y)
    plt.plot(x, np.full_like(y, mean_y, dtype=float), linestyle="-", c="red", label=f"Average RMSD = {mean_y:.2f} ± {std_y:.2f} Å")
    plt.plot(x, np.full_like(y, (mean_y - std_y), dtype=float), linestyle="--", c="red", linewidth=0.5)
    plt.plot(x, np.full_like(y, (mean_y + std_y), dtype=float), linestyle="--", c="red", linewidth=0.5)
    plt.xlabel("Time (ns)", fontsize=12, fontweight="bold")
    plt.ylabel("RMSD (Å)", fontsize=12, fontweight="bold")
    plt.grid(color="0.9", linestyle="-", linewidth=0.5)
    plt.legend()
    
    if file == "rmsd_vs_start.xvg":
        if mutation == False:
            directory = path[-3:].replace("/", "")
            plt.title(f"RMSD for all complex {directory} {prot1}-{prot2}", fontsize=14, fontweight="bold")
            plt.savefig(f"{path}rmsd_plot_{directory}_all_{prot1}vs{prot2}.png", dpi=300)
        elif mutation == True:
            directory = path[-9:-6].replace("/", "")
            name_mut = path[-6:-1]
            plt.title(f"RMSD for all complex {directory} {prot1}_{name_mut}-{prot2}", fontsize=14, fontweight="bold")
            plt.savefig(f"{path}rmsd_plot_{directory}_all_{prot1}_{name_mut}vs{prot2}.png", dpi=300)
        else:
            print(f"{mutation} must be a boolean value (True or False).")
            
    elif file == "rmsd_peptide.xvg":
        if mutation == False:
            directory = path[-3:].replace("/", "")
            plt.title(f"RMSD for {prot1} with {prot2} in {directory}", fontsize=14, fontweight="bold")
            plt.savefig(f"{path}rmsd_plot_{directory}_{prot1}vs{prot2}.png", dpi=300)
        elif mutation == True:
            directory = path[-9:-6].replace("/", "")
            name_mut = path[-6:-1]
            plt.title(f"RMSD for {prot1}_{name_mut} with {prot2} in {directory}", fontsize=14, fontweight="bold")
            plt.savefig(f"{path}rmsd_plot_{directory}_{prot1}_{name_mut}vs{prot2}.png", dpi=300)
        else:
            print(f"{mutation} must be a boolean value (True or False).")
    
    elif file == "rmsd_protein.xvg":
        if mutation == False:
            directory = path[-3:].replace("/", "")   
            plt.title(f"RMSD for {prot2} with {prot1} in {directory}", fontsize=14, fontweight="bold")
            plt.savefig(f"{path}rmsd_plot_{directory}_{prot2}vs{prot1}.png", dpi=300)
        elif mutation == True:
            directory = path[-9:-6].replace("/", "")
            name_mut = path[-6:-1]
            plt.title(f"RMSD for {prot2} with {prot1}_{name_mut} in {directory}", fontsize=14, fontweight="bold")
            plt.savefig(f"{path}rmsd_plot_{directory}_{prot2}vs{prot1}_{name_mut}.png", dpi=300)
        else:
            print(f"{mutation} must be a boolean value (True or False).")

    else:
        print(f"{file} value must be 'rmsd_vs_start.xvg', 'rmsd_peptide.xvg' or 'rmsd_protein.xvg'.")
    plt.show()
```

```python
def rmsf(path, prot1, prot2, mutation=False, name_mut=""):
    """
    """
    file = "rmsf_per_residue.xvg"
    x, y = [], []
    
    if prot1 == "MRD":
        with open(path+file) as f:
            i = 0
            for line in f:
                if line.startswith(("@", "#")):
                    continue
                i += 1
                if i <= 16:
                    cols = line.strip().split()
                    x.append(float(cols[0]))
                    y.append(float(cols[1])*10)
                    
        plt.figure(figsize=(12, 6))
        plt.plot(x, y)
        plt.ylabel("RMSF (Å)", fontsize=12, fontweight="bold")
        plt.ylim(0,)
        plt.gca().xaxis.set_major_locator(MultipleLocator(1))
        plt.grid(color="0.9", linestyle="-", linewidth=0.5)
        
        if mutation == False:
            plt.title(f"RMSF for {prot1} in {prot1}-{prot2} complex", fontsize=14, fontweight="bold")
            plt.xlabel(f"{prot1} residue", fontsize=12, fontweight="bold")
            plt.savefig(f"{path}rmsf_plot_{prot1}.png", dpi=300)  
        elif mutation == True:
            plt.title(f"RMSF for {prot1}_{name_mut} in {prot1}-{prot2} complex", fontsize=14, fontweight="bold")
            plt.xlabel(f"{prot1}_{name_mut} residue", fontsize=12, fontweight="bold")
            plt.savefig(f"{path}rmsf_plot_{prot1}_{name_mut}.png", dpi=300)
        else:
            print(f"{mutation} must be a boolean value (True or False).")
        plt.show()
    else:
        print(f"{prot1} must be 'MRD'.")

    
    x, y = [], []
    with open(path+file) as f:
        i = 0
        for line in f:
            if line.startswith(("@", "#")):
                continue
            i += 1
            if i > 16:
                cols = line.strip().split()
                x.append(float(cols[0]))
                y.append(float(cols[1])*10)

    plt.figure(figsize=(12, 6))
    plt.plot(x, y)
    plt.ylabel("RMSF (Å)", fontsize=12, fontweight="bold")
    plt.ylim(0,)
    plt.xlabel(f"{prot2} residue", fontsize=12, fontweight="bold")
    plt.grid(color="0.9", linestyle="-", linewidth=0.5)

    if mutation == False:
        plt.title(f"RMSF for {prot2} in {prot1}-{prot2} complex", fontsize=14, fontweight="bold")      
        plt.savefig(f"{path}rmsf_plot_{prot2}.png", dpi=300)
    elif mutation == True:
        plt.title(f"RMSF for {prot2} in {prot1}_{name_mut}-{prot2} complex", fontsize=14, fontweight="bold")
        plt.savefig(f"{path}rmsf_plot_{prot2}_{prot1}_{name_mut}.png", dpi=300)
    else:
        print(f"{mutation} must be a boolean value (True or False).")
    plt.show()
```

```python
def rmsf_zoom(path, prot, begin, end, mutation=False, name_mut=""):
    """
    """
    file = "rmsf_per_residue.xvg"
    x, y = [], []
    
    with open(path+file) as f:
        i = 0
        for line in f:
            if line.startswith(("@", "#")):
                continue
            i += 1
            if i > 16:
                cols = line.strip().split()
                x.append(int(cols[0]))
                y.append(float(cols[1])*10)
    
    filtered = []
    i = 0
    
    while i < len(x):
        if begin <= x[i] <= end:
            filtered.append((x[i], y[i]))
        i += 1
        
    filtered = np.array(filtered)
    x_axe = list(range(int(begin), int(end)+1))
    
    plt.figure(figsize=(12, 6))
    plt.plot(filtered[:,0], filtered[:,1])
    plt.ylabel("RMSF (Å)", fontsize=12, fontweight="bold")
    plt.ylim(0,)
    plt.xlabel(f"{prot} residue", fontsize=12, fontweight="bold")
    plt.xticks(x_axe)
    plt.gca().xaxis.set_major_locator(MultipleLocator(1))
    plt.grid(color="0.9", linestyle="-", linewidth=0.5)

    if mutation == False:
        plt.title(f"RMSF between {begin}-{end} for {prot} in the WT complex", fontsize=14, fontweight="bold")      
        plt.savefig(f"{path}rmsf_plot_{prot}_{begin}_{end}.png", dpi=300) 
    elif mutation == True:
        plt.title(f"RMSF between {begin}-{end} for {prot} in the {name_mut} complex", fontsize=14, fontweight="bold")
        plt.savefig(f"{path}rmsf_plot_{prot}_{begin}_{end}_with_{name_mut}.png", dpi=300)
    else:
        print(f"{mutation} must be a boolean value (True or False).")
    plt.show()
```

```python
def averaged_rmsd(path, prot1, prot2, file, mutation, name_mut):
    """
    """
    x = []
    y1, y2, y3 = [], [], []
    
    if mutation == False:
        directory = path[:-3]
        with open(directory+"V1/"+file) as f1:
            
            for line in f1:
                if line.startswith(("@", "#")):
                    continue
                cols1 = line.strip().split()
                x.append(float(cols1[0])/1000) # time in ns
                y1.append(float(cols1[1])*10) # distance in Å
            
        mean_y1 = np.mean(y1)
        std_y1 = np.std(y1)
        
        with open(directory+"V2/"+file) as f2:
            
            for line in f2:
                if line.startswith(("@", "#")):
                    continue
                cols2 = line.strip().split()
                y2.append(float(cols2[1])*10) # distance in Å

        mean_y2 = np.mean(y2)
        std_y2 = np.std(y2)
        
        with open(directory+"V3/"+file) as f3:
            
            for line in f3:
                if line.startswith(("@", "#")):
                    continue
                cols3 = line.strip().split()
                y3.append(float(cols3[1])*10) # distance in Å

        mean_y3 = np.mean(y3)
        std_y3 = np.std(y3)
    
    elif mutation == True:
        directory = path[:-9]
        with open(directory+f"V1/{name_mut}/"+file) as f1:
            
            for line in f1:
                if line.startswith(("@", "#")):
                    continue
                cols1 = line.strip().split()
                x.append(float(cols1[0])/1000) # time in ns
                y1.append(float(cols1[1])*10) # distance in Å
        
        mean_y1 = np.mean(y1)
        std_y1 = np.std(y1)
        
        with open(directory+f"V2/{name_mut}/"+file) as f2:
            
            for line in f2:
                if line.startswith(("@", "#")):
                    continue
                cols2 = line.strip().split()
                y2.append(float(cols2[1])*10) # distance in Å

        mean_y2 = np.mean(y2)
        std_y2 = np.std(y2)
        
        with open(directory+f"V3/{name_mut}/"+file) as f3:
            for line in f3:
                if line.startswith(("@", "#")):
                    continue
                cols3 = line.strip().split()
                y3.append(float(cols3[1])*10) # distance in Å

        mean_y3 = np.mean(y3)
        std_y3 = np.std(y3)

    else:
        print(f"{mutation} must be a boolean value (True or False).")
        return None

    averaged_y = [(a + b + c)/3 for a, b, c in zip(y1, y2, y3)]
    averaged_mean = (mean_y1 + mean_y2 + mean_y3)/3
    averaged_std = (std_y1 + std_y2 + std_y3)/3

    return x, averaged_y, averaged_mean, averaged_std, directory
            
```

```python
def plot_averaged_rmsd(path, prot1, prot2, file, mutation=False, name_mut=""):
    """
    """

    # For RMSD analyse
    if file == "rmsd_vs_start.xvg" or "rmsd_peptide.xvg" or "rmsd_protein.xvg":
        if file == "rmsd_peptide.xvg":
            title = f"{prot1}"
        elif file == "rmsd_protein.xvg":
            title = f"{prot2}"
        else:
            title = "all"

        parameter_y = "RMSD"
        parameter_x = "Time (ns)"
        x, averaged, averaged_mean, averaged_std, directory = averaged_rmsd(path, prot1, prot2, file, mutation, name_mut)
            
        # Plot
        plt.figure(figsize=(12, 6))
        plt.plot(x, averaged)
        plt.plot(x, np.full_like(averaged, averaged_mean, dtype=float), linestyle="-", c="red", label=f"Average {parameter_y} = {averaged_mean:.2f} ± {averaged_std:.2f} Å")
        plt.plot(x, np.full_like(averaged, (averaged_mean - averaged_std), dtype=float), linestyle="--", c="red", linewidth=0.5)
        plt.plot(x, np.full_like(averaged, (averaged_mean + averaged_std), dtype=float), linestyle="--", c="red", linewidth=0.5)
        plt.xlabel(f"{parameter_x}", fontsize=12, fontweight="bold")
        plt.ylabel(f"{parameter_y} (Å)", fontsize=12, fontweight="bold")
        plt.grid(color="0.9", linestyle="-", linewidth=0.5)
        plt.legend()
        if mutation == False:
            plt.title(f"Average {parameter_y} for {title} in complex {prot1}-{prot2}", fontsize=14, fontweight="bold")
            plt.savefig(f"{directory}average_{parameter_y}_plot_{title}_in_{prot1}vs{prot2}.png", dpi=300)
        else:
            plt.title(f"Average {parameter_y} for {title} in complex {prot1}_{name_mut}-{prot2}", fontsize=14, fontweight="bold")
            plt.savefig(f"{directory}average_{parameter_y}_plot_{title}_in_{prot1}_{name_mut}vs{prot2}.png", dpi=300)
        plt.show()

    else:
        print(f"{file} must be a RMSD file (XVG format).")
        return None
```

```python
u_gro = mda.Universe(GRO, XTC)
#u_pdb = mda.Universe(PDB, XTC)
```

```python
# Fonction qui trouve le nombre d'atomes dans le peptide associé au WD40
# AUTO: Trouver la première ligne (fichier pdb et gro)
# Fichier spé? len sequence des prot A et B?

def find_atoms_numbers(gro_file):
    
    with open(gro_file, "r") as f:
        lines = f.readlines()
    
    atom_lines = lines[2:-1]
    atoms = 0
    
    for line in atom_lines:
        atoms += 1
        if line.startswith("  166ARG      N"):
            return(int(atoms-1))
```

```python
# ATTENTION: le compte des atomes commencent à 0 pour les fichiers .gro et à 1 pour les fichiers .pdb!
# AUTO: changer le numéros des atomes en fonction du complexe (mNFIL3, mMECP2, mMKX...) => 1er ligne du fichier "output_analyse.out"
# Fichier spé? len sequence des prot A et B?

peptide_count = find_atoms_numbers(base_path+"md_0_1.gro")
print(peptide_count)

protein_begin = peptide_count
protein_end = protein_begin + 5251 # TO DO: ne fonctionne que pour le domaine WD40 de TBL1R!

MRD = u_gro.atoms[0:peptide_count]
print(MRD)
WD40 = u_gro.atoms[protein_begin:protein_end]
print(WD40)
```

```python
# Définition des pseudo ponts-salins comme contacts
# Basé sur? Pourquoi? Autres méthodes? AA hydrophobes?

sel_basic = "(resname ARG LYS HIS) and (name NH* NZ ND1 NE NE2 HH1* HH2* HE HE2 HZ*)"
sel_acidic = "name OE* OD* O"

basic = MRD.select_atoms(sel_basic)
print(basic, len(basic))
acidic = WD40.select_atoms(sel_acidic)
print(acidic, len(acidic))
```

```python
### Analyse du nombre de contacts ###
```

```python
# Fonction qui calcule le nombre de contacts entre 2 groupes et dans le rayon définis pour chaque frame de la trajectoire

def contacts_within_cutoff(u, group_a, group_b, radius=4.5):
    timeseries = []
    for ts in u.trajectory:
        # calculate distances between group_a and group_b
        dist = contacts.distance_array(group_a.positions, group_b.positions)
        # determine which distances <= radius
        n_contacts = contacts.contact_matrix(dist, radius).sum()
        timeseries.append([ts.frame, ts.frame/100, n_contacts])
    return np.array(timeseries)
```

```python
# Lent? Amélioration possible?
# + de 30 putain de minutes!

ca = contacts_within_cutoff(u_gro, acidic, basic, radius=4.5)
ca.shape
```

```python
########## Calcul et plot de l'énergie de liaisons Peptide-Proteine au cours du temps ##########
```

```python
"""
Calculate MM binding energy (Electrostatic + VdW) in kJ/mol
Uses simplified Coulomb and Lennard-Jones potentials
"""

print("Calculating MM Binding Energy...")

energies = []
time_ns = []

# Constants
epsilon_0 = 8.854187817e-12  # Vacuum permittivity (F/m)
epsilon_r = 1.0  # Relative permittivity (vacuum approximation)
e = 1.602176634e-19  # Elementary charge (C)
Na = 6.02214076e23  # Avogadro's number

# Coulomb constant in kJ/(mol·nm)
# k_e = (1 / (4π * ε₀ * ε_r)) * e² * Na / 1000
k_e = 138.935458  # kJ·nm/(mol·e²) - standard value for MD simulations

# Lennard-Jones parameters (typical protein-ligand values)
epsilon_lj = 0.5  # kJ/mol - depth of potential well
sigma = 0.35  # nm - distance at which potential is zero

for ts in u_gro.trajectory:
    # Get protein-ligand distance matrix in Angstroms
    distances_angstrom = mda.lib.distances.distance_array(MRD.positions, WD40.positions)

    # Convert to nm for proper energy calculation
    distances_nm = distances_angstrom / 10.0

    # Avoid division by zero
    distances_nm = np.maximum(distances_nm, 0.01)  # Minimum distance 0.01 nm

    # Electrostatic energy (Coulomb's law)
    # E_elec = k_e * q1 * q2 / r
    # Assuming partial charges of ±0.5e for simplified calculation
    q_protein = 0.5  # Partial charge (elementary charge units)
    q_ligand = 0.5
    elec_energy = k_e * q_protein * q_ligand / distances_nm
    total_elec = np.sum(elec_energy)

    # Van der Waals energy (Lennard-Jones 6-12 potential)
    # E_vdw = 4ε[(σ/r)^12 - (σ/r)^6]
    sigma_r = sigma / distances_nm
    vdw_energy = 4 * epsilon_lj * (sigma_r**12 - sigma_r**6)
    total_vdw = np.sum(vdw_energy)

    # Total interaction energy
    total_energy = total_elec + total_vdw

    energies.append(total_energy)
    time_ns.append(ts.time / 1000.0)

print("Calcul MM Binding Energy done!")
```

```python
"""
Plot binding energy over time in kJ/mol
"""

fig, ax = plt.subplots(figsize=(12, 6))
ax.plot(time_ns, energies)
ax.set_xlabel("Time (ns)", fontsize=12, fontweight="bold")
ax.set_ylabel("Interaction Energy (kJ/mol)", fontsize=12, fontweight="bold")
ax.set_title(f"{peptide}-{protein} Interaction Energy", fontsize=14, fontweight="bold")
ax.grid(alpha=0.5)

mean_energy = np.mean(energies)
std_energy = np.std(energies)
ax.axhline(mean_energy, c="red", linestyle="-", label=f"Mean: {mean_energy:.2f} kJ/mol")
ax.axhline((mean_energy - std_energy), c="red", linestyle="--", linewidth=0.5)
ax.axhline((mean_energy + std_energy), c="red", linestyle="--", linewidth=0.5)
ax.legend()

plt.tight_layout()
plt.savefig(f"{base_path}energy.png", dpi=300)
print(f"Binding energy plot saved to {base_path}energy.png")
print(f"Mean Energy: {mean_energy:.2f} ± {std_energy:.2f} kJ/mol")
plt.show()
```

```python
"""Calculate RMSD over trajectory"""

self = u_gro
selection="backbone"
reference_frame=0

print("Calculating RMSD...")

if selection == "backbone":
    sel = self.select_atoms("backbone")
elif selection == "name CA":
    sel = self.select_atoms("name CA")
else:
    sel = self.select_atoms(selection)

R = rms.RMSD(sel, sel, select=selection, ref_frame=reference_frame)
R.run()

# Convert time to ns
time_ns = R.results.rmsd[:, 1] / 1000.0
rmsd_values = R.results.rmsd[:, 2]

print("Calcul RMSD done!")
```

```python
"""Plot RMSD vs Time"""

fig, ax = plt.subplots(figsize=(12, 6))
ax.plot(time_ns, rmsd_values)
ax.set_title("Backbone RMSD over Time", fontsize=14, fontweight="bold")
ax.set_xlabel("Time (ns)", fontsize=12, fontweight="bold")
ax.set_ylabel("RMSD (Å)", fontsize=12, fontweight="bold")
ax.grid(alpha=0.3)

# Add statistics - only mean line
mean_rmsd = np.mean(rmsd_values)
std_rmsd = np.std(rmsd_values)
ax.axhline(mean_rmsd, color="red", linestyle="-", alpha=0.9, label=f"Mean: {mean_rmsd:.2f} ± {std_rmsd:.2f} Å")
ax.axhline(mean_rmsd - std_rmsd, color="red", linestyle="--", alpha=0.6)
ax.axhline(mean_rmsd + std_rmsd, color="red", linestyle="--", alpha=0.6)
ax.legend()

plt.tight_layout()
plt.savefig(f"{base_path}rmsd.png", dpi=300)
print(f"RMSD plot saved to {base_path}rmsd.png")
plt.show()
```
